```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# 4 Modalities

```python

# === Config ===
output_dir = "/data/wang/junh/results/Fuse_moe/all_los/multiclass/kl_mi/"  # Replace with your output directory
missingInd = False              # Set to True if missing modality was used
modeltype = "TS_CXR_Text_ECG"         # Replace manually as needed
weight_type = "kl_divergence"   # or "correlation_weight", "mutual_weight"

# === Collect CSVs and Concatenate ===
all_dfs = []
for epoch in range(8):  # Replace with the actual number of epochs
    if missingInd:
        fname = f"{output_dir}/Missing_{modeltype}_{epoch}_weights_log.csv"
    else:
        fname = f"{output_dir}/{modeltype}_{epoch}_weights_log.csv"
    
    if os.path.exists(fname):
        df = pd.read_csv(fname)
        df['epoch'] = epoch  # ensure epoch is int for sorting
        all_dfs.append(df)

# Combine all epochs
df_all = pd.concat(all_dfs, ignore_index=True)

# === Plotting ===
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_all, x="epoch", y=df_all['kl_divergence'], hue="modality", marker='o')

plt.title(f"KL * MI Over Epochs (subset)")
plt.xlabel("Epoch")
plt.ylabel("KL * MI")
plt.legend(title="Modality")
plt.tight_layout()
plt.show()
```
